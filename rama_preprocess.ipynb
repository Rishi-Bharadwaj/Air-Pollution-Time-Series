```python
import os
import zipfile

zip_dir   = "/home/rishi/ML Projects/Air Pollution/Datasets/rama_data"
unzip_dir = "/home/rishi/ML Projects/Air Pollution/RAMA/raw"
start_year = 2022
end_year   = 2025

os.makedirs(unzip_dir, exist_ok=True)

year_suffixes = {str(y)[-2:] for y in range(start_year, end_year + 1)}

extracted = 0
for fname in os.listdir(zip_dir):
    if not fname.endswith(".zip"):
        continue
    if "RAMA" not in fname:
        continue
    name_no_ext = fname[:-4]
    if not any(name_no_ext.startswith(s) for s in year_suffixes):
        continue
    fpath = os.path.join(zip_dir, fname)
    with zipfile.ZipFile(fpath, "r") as zf:
        zf.extractall(unzip_dir)
    extracted += 1
    print(f"Extracted: {fname}")

print(f"Done. Extracted {extracted} zip files to {unzip_dir}")
```

```python
import pandas as pd

converted, deleted = 0, 0
for fname in os.listdir(unzip_dir):
    if not fname.endswith(".xls"):
        continue
    fpath = os.path.join(unzip_dir, fname)
    df = pd.read_excel(fpath)
    csv_path = os.path.join(unzip_dir, fname[:-4] + ".csv")
    df.to_csv(csv_path, index=False)
    os.remove(fpath)
    converted += 1

print(f"Done. Converted {converted} .xls files to CSV in {unzip_dir}")
```
