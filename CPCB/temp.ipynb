```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


def plot_site_comparison_heatmap(df, feature_name):
    nan_counts = df.isnull().sum().sort_values()
    df = df.reindex(columns=nan_counts.index)
    df_transposed = df.T

    # Convert to binary: 0 -> 0 (black), non-zero -> 1 (white), NaN stays masked
    binary_data = np.where(df_transposed.isna(), np.nan, np.where(df_transposed == 0, 0, 1))

    fig, ax = plt.subplots(figsize=(20, len(df.columns) * 0.025))

    sns.heatmap(
        binary_data,
        cmap='gray',
        cbar_kws={'label': '0 (black) vs non-zero (white)'},
        xticklabels=False,
        yticklabels=False,
        ax=ax,
        mask=np.isnan(binary_data),
        vmin=0,
        vmax=1
    )

    # Add yearly ticks
    years = pd.to_datetime(df.index).year.unique()
    year_starts = [df.index.get_loc(str(year) + '-01-01 00:00:00') for year in years if str(year) + '-01-01 00:00:00' in df.index]
    ax.set_xticks(year_starts)
    ax.set_xticklabels(years, rotation=0, fontsize=10)

    plt.title(f'{feature_name} Zero vs Non-Zero Across Sites Over Time', fontsize=16, pad=20)
    plt.xlabel('Year', fontsize=12)
    plt.ylabel('Site', fontsize=12)
    plt.tight_layout()
    plt.show()
```

```python

feature_df=pd.read_csv(rf"/home/rishi/ML Projects/Air Pollution/CPCB/visualize_median_2023/CO_(mg_m³)_df_cpcb_new_limits.csv", index_col=0, parse_dates=True)
plot_site_comparison_heatmap(feature_df,feature_name="CO")
```
